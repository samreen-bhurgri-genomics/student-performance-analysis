```python
import os

os.path.exists("StudentsPerformance_Cleaned.csv")
```

```python
clean_df = pd.read_csv("StudentsPerformance_Cleaned.csv")

print(clean_df.head())
```

```python
print("Original dataset:", df.shape)
print("Cleaned dataset:", clean_df.shape)
```

```python
print("Average Math Score:", clean_df["math score"].mean())
print("Average Reading Score:", clean_df["reading score"].mean())
print("Average Writing Score:", clean_df["writing score"].mean())
```

```python
gender_avg = clean_df.groupby("gender")[["math score" , "reading score", "writing score"]].mean()
print(gender_avg)
```

```python
subject_avg =clean_df[["math score", "reading score", "writing score"]].mean()
print(subject_avg)
print("Highest average subject:" , subject_avg.idxmax())
```

```python
import matplotlib.pyplot as plt

subject_avg.plot(kind="bar")

plt.title("Average Marks by subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

bars = subject_avg.plot(
    kind="bar",
    color=["skyblue", "orange", "lightgreen"]
)

plt.title("Average Student Marks by Subject", fontsize=14)
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.ylim(0, 100)

plt.bar_label(bars.containers[0], fmt="%.2f")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
high_math = clean_df[clean_df["math score"] > 80]

print("Students scoring above 80 in Math:", len(high_math))
```

```python
high_math = [clean_df["math score"]>80]
print("Students scoring above 80 in Math:",len(high_math))
```

```python
low_math = clean_df[clean_df["math score"] < 40]
print("Students scoring below 40:" , len(low_math))
```

```python
prep_avg = clean_df.groupby("test preparation course")[["math score", "reading score", "writing score"]].mean()
print(prep_avg)
```

```python
prep_avg.plot(kind="bar", figsize=(9, 5))

plt.title("Average Marks: Test Preparation Course")
plt.xlabel("Test Preparation")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.legend(title="Subjects")
plt.tight_layout()
plt.show()
```

```python
print("STUDENT PERFORMANCE ANALYSIS")
print("----------------------------")

print("Total Students:", len(clean_df))
print("Highest Average Subject:", subject_avg.idxmax())
print("Highest Math Average:", gender_avg["math score"].idxmax())
print("Highest Reading Average:", gender_avg["reading score"].idxmax())
print("Highest Writing Average:", gender_avg["writing score"].idxmax())
print("Best Preparation Group in Math:", prep_avg["math score"].idxmax())
```

```python
report = pd.DataFrame({
    "Subject": ["Math", "Reading", "Writing"],
    "Average Marks": [
        subject_avg["math score"],
        subject_avg["reading score"],
        subject_avg["writing score"]
    ]
})

report.to_csv("Student_Analysis_Report.csv", index=False)

print("Report saved successfully!")
```
